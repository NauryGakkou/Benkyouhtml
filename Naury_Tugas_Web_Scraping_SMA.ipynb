```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. Tentukan URL target web scraping
url = "https://chomsky.info/audionvideo/"

# 2. Kirim HTTP request ke website
response = requests.get(url)

# 3. Cek apakah request berhasil (status code 200)
if response.status_code == 200:
    # 4. Parsing HTML menggunakan BeautifulSoup
    soup = BeautifulSoup(response.text, "html.parser")

    # 5. Ambil data target (contoh: mengambil semua teks dari tag <li>)
    items = soup.find_all("li")

    data = []
    for item in items:
        text = item.get_text(strip=True)
        if text:  # Memastikan teks tidak kosong
            data.append(text)

    # 6. Masukkan data ke dalam DataFrame Pandas
    df = pd.DataFrame(data, columns=["Judul / Informasi"])

    # Tampilkan 5 data teratas di Colab
    display(df.head())

    # 7. Simpan hasil ke file CSV
    df.to_csv("hasil_scraping_chomsky.csv", index=False)
    print("\nScraping Berhasil! File CSV telah disimpan.")
else:
    print(f"Gagal mengakses halaman. Status code: {response.status_code}")
```
